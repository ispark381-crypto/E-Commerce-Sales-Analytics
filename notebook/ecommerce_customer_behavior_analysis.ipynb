```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = "../dataset/ecommerce_customer_behavior_dataset_v2.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

print("Date converted successfully!")
print(df['Date'].dtype)
print("Invalid dates:", df['Date'].isnull().sum())
```

```python
# Create a cleaned copy
df_clean = df.copy()

# Sort records by date
df_clean = df_clean.sort_values('Date').reset_index(drop=True)

print("Cleaned dataset created successfully!")
print("Shape:", df_clean.shape)
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
cleaned_path = "../dataset/ecommerce_customer_behavior_cleaned.csv"

df_clean.to_csv(cleaned_path, index=False)

print("Cleaned dataset saved successfully!")
print(cleaned_path)
```

```python
category_sales = (
    df_clean.groupby('Product_Category')['Total_Amount']
    .sum()
    .sort_values(ascending=False)
    .round(2)
)

display(category_sales)
```

```python
payment_sales = (
    df_clean.groupby('Payment_Method')['Total_Amount']
    .sum()
    .sort_values(ascending=False)
    .round(2)
)

display(payment_sales)
```

```python
device_sales = (
    df_clean.groupby('Device_Type')['Total_Amount']
    .sum()
    .sort_values(ascending=False)
    .round(2)
)

display(device_sales)
```

```python
monthly_sales = (
    df_clean.groupby(df_clean['Date'].dt.to_period('M'))['Total_Amount']
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

display(monthly_sales.round(2))
```

```python
correlation_matrix = df_clean[
    [
        'Age',
        'Unit_Price',
        'Quantity',
        'Discount_Amount',
        'Total_Amount',
        'Session_Duration_Minutes',
        'Pages_Viewed',
        'Delivery_Time_Days',
        'Customer_Rating'
    ]
].corr().round(2)

display(correlation_matrix)
```

```python
numeric_columns = [
    'Age',
    'Unit_Price',
    'Quantity',
    'Discount_Amount',
    'Total_Amount',
    'Session_Duration_Minutes',
    'Pages_Viewed',
    'Delivery_Time_Days',
    'Customer_Rating'
]

outlier_summary = {}

for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()

    outlier_summary[col] = count

outlier_summary = pd.Series(outlier_summary).sort_values(ascending=False)

display(outlier_summary)
```

```python
delivery_by_category = (
    df_clean.groupby('Product_Category')['Delivery_Time_Days']
    .mean()
    .sort_values()
)

plt.figure(figsize=(10, 6))

delivery_by_category.plot(kind='barh')

plt.title('Average Delivery Time by Product Category')
plt.xlabel('Average Delivery Time (Days)')
plt.ylabel('Product Category')

plt.tight_layout()
plt.savefig('../visualizations/average_delivery_time_by_category.png', dpi=300)
plt.show()
```

```python
# Create a compact Power BI-ready summary dataset

power_bi_data = []

# KPI metrics
power_bi_data.extend([
    ['KPI', 'Overall', 'Total Sales', df_clean['Total_Amount'].sum()],
    ['KPI', 'Overall', 'Total Orders', df_clean['Order_ID'].nunique()],
    ['KPI', 'Overall', 'Total Quantity', df_clean['Quantity'].sum()],
    ['KPI', 'Overall', 'Average Rating', df_clean['Customer_Rating'].mean()],
    ['KPI', 'Overall', 'Average Delivery Time', df_clean['Delivery_Time_Days'].mean()]
])

# Product category analysis
for category, group in df_clean.groupby('Product_Category'):
    power_bi_data.append(
        ['Product Category', category, 'Sales', group['Total_Amount'].sum()]
    )

# Payment method analysis
for payment, group in df_clean.groupby('Payment_Method'):
    power_bi_data.append(
        ['Payment Method', payment, 'Sales', group['Total_Amount'].sum()]
    )

# Device type analysis
for device, group in df_clean.groupby('Device_Type'):
    power_bi_data.append(
        ['Device Type', device, 'Sales', group['Total_Amount'].sum()]
    )

# Monthly sales analysis
for month, group in df_clean.groupby(df_clean['Date'].dt.to_period('M')):
    power_bi_data.append(
        ['Month', str(month), 'Sales', group['Total_Amount'].sum()]
    )

# Delivery time by category
for category, group in df_clean.groupby('Product_Category'):
    power_bi_data.append(
        ['Delivery Time', category, 'Average Delivery Days',
         group['Delivery_Time_Days'].mean()]
    )

# Convert to DataFrame
power_bi_df = pd.DataFrame(
    power_bi_data,
    columns=['Analysis_Type', 'Category', 'Metric', 'Value']
)

# Round values
power_bi_df['Value'] = power_bi_df['Value'].round(2)

# Display
display(power_bi_df)

# Save for Power BI
power_bi_path = "../dataset/power_bi_dashboard_data.csv"
power_bi_df.to_csv(power_bi_path, index=False)

print("\nPower BI file created successfully:")
print(power_bi_path)
print("Rows:", len(power_bi_df))
```
