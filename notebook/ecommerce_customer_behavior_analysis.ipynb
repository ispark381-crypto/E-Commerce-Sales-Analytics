```python
import pandas as pd
import numpy as np

file_path = "../dataset/ecommerce_customer_behavior_dataset_v2.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

print("Date converted successfully!")
print(df['Date'].dtype)
print("Invalid dates:", df['Date'].isnull().sum())
```

```python
print("Age:", df['Age'].min(), "to", df['Age'].max())
print("Quantity:", df['Quantity'].min(), "to", df['Quantity'].max())
print("Unit Price:", df['Unit_Price'].min(), "to", df['Unit_Price'].max())
print("Discount Amount:", df['Discount_Amount'].min(), "to", df['Discount_Amount'].max())
print("Total Amount:", df['Total_Amount'].min(), "to", df['Total_Amount'].max())
print("Session Duration:", df['Session_Duration_Minutes'].min(), "to", df['Session_Duration_Minutes'].max())
print("Pages Viewed:", df['Pages_Viewed'].min(), "to", df['Pages_Viewed'].max())
print("Delivery Time:", df['Delivery_Time_Days'].min(), "to", df['Delivery_Time_Days'].max())
print("Customer Rating:", df['Customer_Rating'].min(), "to", df['Customer_Rating'].max())

print("\nProduct Categories:")
print(df['Product_Category'].unique())

print("\nPayment Methods:")
print(df['Payment_Method'].unique())

print("\nDevice Types:")
print(df['Device_Type'].unique())
```

```python
# Create a cleaned copy
df_clean = df.copy()

# Sort records by date
df_clean = df_clean.sort_values('Date').reset_index(drop=True)

print("Cleaned dataset created successfully!")
print("Shape:", df_clean.shape)
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
cleaned_path = "../dataset/ecommerce_customer_behavior_cleaned.csv"

df_clean.to_csv(cleaned_path, index=False)

print("Cleaned dataset saved successfully!")
print(cleaned_path)
```

```python
monthly_sales = (
    df_clean.groupby(df_clean['Date'].dt.to_period('M'))['Total_Amount']
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

display(monthly_sales.round(2))
```

```python
correlation_matrix = df_clean[
    [
        'Age',
        'Unit_Price',
        'Quantity',
        'Discount_Amount',
        'Total_Amount',
        'Session_Duration_Minutes',
        'Pages_Viewed',
        'Delivery_Time_Days',
        'Customer_Rating'
    ]
].corr().round(2)

display(correlation_matrix)
```

```python
numeric_columns = [
    'Age',
    'Unit_Price',
    'Quantity',
    'Discount_Amount',
    'Total_Amount',
    'Session_Duration_Minutes',
    'Pages_Viewed',
    'Delivery_Time_Days',
    'Customer_Rating'
]

outlier_summary = {}

for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()

    outlier_summary[col] = count

outlier_summary = pd.Series(outlier_summary).sort_values(ascending=False)

display(outlier_summary)
```
